```python
age = [19,22,21,35,28,24,31,19,27,40]
print(len(age))
```

```python
total_age = sum(age)
print (total_age)
```

```python
mean_age = total_age / len(age)
print (mean_age)
```

```python
min_age = min(age)
print (min_age)
max_age = max(age)
print (max_age)
```

```python
import statistics
median_age = statistics.median(age)
print (median_age)
```

Yes the mean 26.6 is different from the median 25.5 ,when the mean and the median disagree, it tells you that the data distribution is skewed rather than perfectly symmetrical
